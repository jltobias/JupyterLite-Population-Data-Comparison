# 1. An invented population comparison

**SYNTHETIC TEACHING DATA:** every bundled count, cell, and allocation is invented. These are not GPW, WorldPop, LandScan, HRSL, GHS-POP, or Esri WPE data. The coordinates locate an arbitrary teaching rectangle and do not assert actual population there.

This notebook runs with Python's standard library in JupyterLite/Pyodide. Use **Run → Run All Cells**. The first kernel start downloads the pinned browser Python runtime; no remote population services are needed. Files are in browser storage; download files you want to keep.

The four scenarios illustrate equal allocation, settlement weighting, a daytime redistribution, and incomplete coverage. Finer cells do not establish higher accuracy. See Leyk et al. (2019), *The spatial allocation of population*, https://doi.org/10.5194/essd-11-1385-2019. The linked-map workflow is inspired by POPGRID Viewer and SEDAC/CIESIN: https://www.earthdata.nasa.gov/news/feature-articles/data-tool-focus-popgrid-viewer.


```python
from population_tools import load_demo, compare, validate_layer
demo = load_demo()
for layer in demo['layers']:
    validate_layer(layer)
    print(layer['metadata']['title'], '| SYNTHETIC |', layer['metadata']['license'])
```

## Recompute the shared example

The rectangle is W=120.1, S=14.1, E=120.3, N=14.3. Its edges coincide with teaching-grid edges. Independently counting the included cells gives **400, 400, 800, and 240** people. The incomplete grid covers only three quarters of this rectangle; a missing cell is not zero population.


```python
results = compare(demo['layers'], demo['default_aoi'])
for result, expected in zip(results, [400, 400, 800, 240]):
    assert abs(result['population'] - expected) < 1e-7
    print(result['metadata']['id'], round(result['population'], 2),
          f"coverage={result['coverage']:.1%}", 'SYNTHETIC')
assert results[-1]['coverage'] < 1
print('BROWSER_COMPUTATION_OK: 400, 400, 800, 240')
```

The fourth coverage is close to 75%, rather than exactly 75%, because a spherical latitude strip's area varies with latitude. Counts use each cell's own spherical area.

Change the rectangle and rerun the next cell. No coverage returns `None`; a covered cell with zero people returns `0.0`.


```python
my_bounds = [120.15, 14.15, 120.25, 14.25]
[(r['metadata']['id'], r['population'], r['coverage'])
 for r in compare(demo['layers'], my_bounds)]
```
